# Advanced Indexing

```python
# import numpy and other dependencies needed for this lab
import numpy as np
```

```python
a =np.arange(12).reshape(4, 3)
a
```

```python
# basic indexing and slicing

print(a[1, 2]) # prints the element at row 1 and column 2
#slicing
print(a[1:3,1:3]) # prints the elements from row 1 to 2 and column 1 to 2
print(a[2:4,0:2]) # prints the elements from row 2 to 3 and column 0 to 1
print(a[::2, ::2]) # prints every second element from row and column
```

# Fancy indexing

```python
a=np.arange(24).reshape(6, 4)
a
```

```python
a[[0,1,3]] # prints the elements from row 0, 1 and 3
a[[0,1,3], [0,1,2]] # prints the elements from row 0, 1 and 3 and columns 0, 1 and 2
a[:, [0,1,2]] # prints the elements from all rows and columns 0, 1 and 2
```

# boolean indexing
Boolean indexing is a technique used to filter or select elements from a NumPy array based on a condition.
Instead of accessing elements using their positions (indexes), we use a Boolean condition that evaluates to True or False.

```python
a = np.random.randint(1, 100, 24).reshape(6, 4)
a
```

```python
a > 50 # this will return a boolean array where each element is True if the corresponding element in 'a' is greater than 50, and False otherwise.
```

```python
#Find all numbers greater than 50
a[a>50] # this will return a 1D array containing only the elements of 'a' that are greater than 50.
```

```python
# Find all numbers greater than 50 and divisible by 2
a
a[(a>50) & (a%2==0)] # this will return a 1D array containing only the elements of 'a' that are greater than 50 and divisible by 2.
```
